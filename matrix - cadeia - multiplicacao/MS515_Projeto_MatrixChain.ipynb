```python
import time

class MultiplicadorDeCadeiaDeMatrizes:
    def __init__(self, dimensoes):
        assert len(dimensoes) > 1
        self.dimensoes = dimensoes
        self.quantidade = len(dimensoes) - 1

    def calcular_operacoes_otimizadas(self):
        # Inicializa a tabela de custos
        custos = [[0] * self.quantidade for _ in range(self.quantidade)]

        # Preenche a tabela em ordem crescente de tamanho da subcadeia
        for comprimento in range(2, self.quantidade + 1):  # comprimento da cadeia
            for i in range(self.quantidade - comprimento + 1):
                j = i + comprimento - 1
                custos[i][j] = float('inf')
                for k in range(i, j):
                    # Calcula o custo de dividir em (A[i..k]) e (A[k+1..j])
                    custo = (
                        custos[i][k]
                        + custos[k + 1][j]
                        + self.dimensoes[i] * self.dimensoes[k + 1] * self.dimensoes[j + 1]
                    )
                    # Atualiza o custo mínimo
                    if custo < custos[i][j]:
                        custos[i][j] = custo

        # Retorna o número mínimo de operações
        return custos[0][self.quantidade - 1]

def main():
    # Solicita ao usuário as dimensões das matrizes
    dimensoes = list(map(int, input("Dimensões: ").split()))

    start_time = time.time()

    # Instancia o objeto para o cálculo
    multiplicador = MultiplicadorDeCadeiaDeMatrizes(dimensoes)

    # Calcula o número total de operações no caso otimizado
    total_operacoes_otimizadas = multiplicador.calcular_operacoes_otimizadas()

    end_time = time.time()
    execution_time = end_time - start_time

    # Exibe o resultado
    print(f"Número mínimo de operações de multiplicação: {total_operacoes_otimizadas}")
    print(f"Tempo de execução: {execution_time:.6f} segundos")

if __name__ == "__main__":
    main()
```

```python
import time

class MultiplicadorDeCadeiaDeMatrizes:
    def __init__(self, dimensoes):
        assert len(dimensoes) > 1
        self.dimensoes = dimensoes

    def calcular_operacoes(self):
        total_operacoes = 0
        for i in range(len(self.dimensoes) - 2):
            # Cálculo de multiplicação de duas matrizes adjacentes
            operacoes = self.dimensoes[i] * self.dimensoes[i + 1] * self.dimensoes[i + 2]
            total_operacoes += operacoes
        return total_operacoes

    def calcular_custo_otimo(self):
        n = len(self.dimensoes) - 1
        # matriz de custos
        custos = [[0 for _ in range(n)] for _ in range(n)]

        # matriz de divisores para reconstrução da ordem
        divisores = [[0 for _ in range(n)] for _ in range(n)]

        # Preenchendo a matriz de custos
        for comprimento in range(2, n + 1):  # comprimento da cadeia
            for i in range(n - comprimento + 1):
                j = i + comprimento - 1
                custos[i][j] = float('inf')  # Inicia com valor muito alto
                for k in range(i, j):
                    q = custos[i][k] + custos[k + 1][j] + self.dimensoes[i] * self.dimensoes[k + 1] * self.dimensoes[j + 1]
                    if q < custos[i][j]:
                        custos[i][j] = q
                        divisores[i][j] = k

        return custos[0][n - 1], divisores

    def reconstruir_ordem(self, divisores, i, j):
        if i == j:
            return f"M{i+1}"
        else:
            k = divisores[i][j]
            esquerda = self.reconstruir_ordem(divisores, i, k)
            direita = self.reconstruir_ordem(divisores, k + 1, j)
            return f"({esquerda} x {direita})"

def main():
    # Solicita ao usuário as dimensões das matrizes
    dimensoes = list(map(int, input("Dimensões (separadas por espaço): ").split()))

    start_time = time.time()

    # Instancia o objeto para o cálculo
    multiplicador = MultiplicadorDeCadeiaDeMatrizes(dimensoes)

    # Calcula o número total de operações diretas
    total_operacoes = multiplicador.calcular_operacoes()

    # Calcula o número total de operações ótimas
    custo_otimo, divisores = multiplicador.calcular_custo_otimo()

    end_time = time.time()
    execution_time = end_time - start_time

    # Exibe os resultados
    print(f"Número total de operações de multiplicação direta: {total_operacoes}")
    print(f"Custo ótimo de multiplicação: {custo_otimo}")
    print(f"Ordem ótima de multiplicação: {multiplicador.reconstruir_ordem(divisores, 0, len(dimensoes) - 2)}")
    print(f"Tempo de execução: {execution_time:.6f} segundos")

if __name__ == "__main__":
    main()
```

```python
import torch
import torch.nn as nn
import torch.optim as optim
from torchvision import datasets, transforms
import time

# Preparação de dados e modelo
transform = transforms.Compose([transforms.ToTensor(), transforms.Normalize((0.5,), (0.5,))])
train_dataset = datasets.MNIST(root='./data', train=True, download=True, transform=transform)
train_loader = torch.utils.data.DataLoader(train_dataset, batch_size=64, shuffle=True)

class RedeNeuralSimples(nn.Module):
    def __init__(self):
        super(RedeNeuralSimples, self).__init__()
        self.fc1 = nn.Linear(28*28, 128)
        self.fc2 = nn.Linear(128, 10)

    def forward(self, x):
        x = torch.flatten(x, 1)
        x = torch.relu(self.fc1(x))
        x = self.fc2(x)
        return x

# Função de treinamento sem otimização
def treinar_sem_otimizacao(modelo, criterio, otimizador, epochs=1):
    modelo.train()
    for epoch in range(epochs):
        running_loss = 0.0
        for inputs, labels in train_loader:
            inputs, labels = inputs.to(device), labels.to(device)
            otimizador.zero_grad()
            outputs = modelo(inputs)
            loss = criterio(outputs, labels)
            loss.backward()
            otimizador.step()
            running_loss += loss.item()
        print(f"Epoch {epoch+1}, Loss: {running_loss/len(train_loader)}")
```

```python
def treinar_com_otimizacao(modelo, criterio, otimizador, dimensoes, epochs=1):
    # Usar o multiplicador de cadeia de matrizes para otimizar
    multiplicador = MultiplicadorDeCadeiaDeMatrizes(dimensoes)
    multiplicador.calcular_custos()

    print(f"Custo ótimo de multiplicação: {multiplicador.get_custo_total()} operações")

    modelo.train()
    for epoch in range(epochs):
        running_loss = 0.0
        for inputs, labels in train_loader:
            inputs, labels = inputs.to(device), labels.to(device)
            otimizador.zero_grad()
            outputs = modelo(inputs)
            loss = criterio(outputs, labels)
            loss.backward()
            otimizador.step()
            running_loss += loss.item()
        print(f"Epoch {epoch+1}, Loss: {running_loss/len(train_loader)}")

# Preparação para execução
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
modelo = RedeNeuralSimples().to(device)
criterio = nn.CrossEntropyLoss()
otimizador = optim.SGD(modelo.parameters(), lr=0.01)
dimensoes = [28*28, 128, 10]  # Exemplo de dimensões de camadas (entrada -> camadas ocultas -> saída)

# Treinamento sem otimização
start_time = time.time()
treinar_sem_otimizacao(modelo, criterio, otimizador, epochs=3)
end_time = time.time()
print(f"Tempo de execução sem otimização: {end_time - start_time:.4f} segundos")

# Treinamento com otimização
start_time = time.time()
treinar_com_otimizacao(modelo, criterio, otimizador, dimensoes, epochs=3)
end_time = time.time()
print(f"Tempo de execução com otimização: {end_time - start_time:.4f} segundos")
```
